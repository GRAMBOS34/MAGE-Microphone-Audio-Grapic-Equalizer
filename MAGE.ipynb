# Basic Audio Playback

At the most fundamental level, the audio engine should be able to play audio. This snippet shows that MAGE can do audio playback. This feature is also important for audio monitoring when using the mic.

Feel free to switch any of the files listed in the "*file*" variable to any of these strings:

| **Song File String** | **Original Song** |
| --------- | ------------- |
| "test_audio/wav/Dracula.wav"     | "*Dracula*" by Tame Impala |
| "test_audio/wav/PlasticLove.wav" | "*Plastic Love*" by Mariya Takeuchi |
| "test_audio/wav/ThankYouForTheVenom.wav" | "*Thank You For The Venom*" by My Chemical Romance |

```python
import include.MAGE as Audio

audioEngine = Audio.MAGE()

file = "test_audio/wav/Dracula.wav" # Plays Dracula - Tame Impala

audioEngine.start_audio_playback(filepath=file)
```

However, this comes with the caveat that it only takes in .wav files. The reason for this is that other file formats (such as .mp3) require an extra layer to turn directly into raw data. .wav files are already in an array that is readable for NumPy and so, it was the easiest to implement. However, adding the extra layer to make .mp3 files readable is quite easy given the structure of the code itself.
